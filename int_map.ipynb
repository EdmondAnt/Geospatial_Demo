```python
import pandas as pd
import folium
```

```python
data = pd.read_csv('Users.csv')
```

```python
data.head(1)
```

```python
geo_data = data[["Country","UserID"]].groupby(by = data["Country"]).agg(total_usr = ("UserID", 'count'))
```

```python
geo_data = geo_data.reset_index()
```

```python
geo_data.head(5)
```

```python
world_map_json = 'https://raw.githubusercontent.com/python-visualization/folium/main/examples/data/world-countries.json'
```

```python
# Create a base world map
m = folium.Map(location=[20, 0], zoom_start=2, tiles="CartoDB positron")
```

```python
folium.Choropleth(
    geo_data=world_map_json,
    name="choropleth",
    data=geo_data,
    columns=["Country", "total_usr"],
    key_on="feature.properties.name",  # Maps your country names to GeoJSON names
    fill_color="YlOrRd",  # Colour scale: Yellow -> Orange -> Red
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name="Number of User IDs",
).add_to(m)

folium.LayerControl().add_to(m)
```

```python
m.save("country_user_heatmap.html")
```

```python
import plotly.express as px
```

```python
fig = px.choropleth(
    geo_data,
    locations="Country",  # Country column
    locationmode="country names",  # Match by standard country names
    color="total_usr",  # Column to shade by
    color_continuous_scale="Reds",  # Heatmap color scheme
    title="User Distribution by Country",
)

# Open directly inside Jupyter notebook/EDA cell
fig.show(renderer="browser")
```
